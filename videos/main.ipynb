```python
import json

with open("remocn_components.json", "r", encoding="utf-8") as f:
    components = json.load(f)
```

```python
import re


def camel_to_kebab(s: str) -> str:
    """
    Convert a camelCase or PascalCase string to kebab-case.
    Handles consecutive uppercase letters as a single word (like "HTTPServer" -> "http-server").
    """
    # Insert hyphen before each uppercase letter that is either:
    # - preceded by a lowercase letter
    # - preceded by an uppercase letter but followed by a lowercase letter (handles acronyms)
    s = re.sub(r"(?<=[a-z])(?=[A-Z])|(?<=[A-Z])(?=[A-Z][a-z])", "-", s)
    return s.lower()
```

```python
from tqdm import tqdm

commands = []

for component in tqdm(components):
    name = camel_to_kebab(component["name"])
    command = f"pnpm dlx shadcn add @remocn/{name}"
    commands.append(command)
```

```python
import subprocess

from tqdm import tqdm

for command in tqdm(commands):
    print(f"Running command: {command}")
    result = subprocess.run(
        command,
        shell=True,  # 如果你确认命令是安全的且需要 shell 特性（如管道、通配符）
        capture_output=True,
        text=True,
        encoding="utf-8",  # 强制使用 UTF-8 解码子进程的输出
    )
    if result.returncode != 0:
        print(f"Error running command: {result.stderr}")
    else:
        print(f"Command output: {result.stdout}")
```
